# Model 5 — Chatbot (RAG Pipeline, Kerala)

This notebook demonstrates a **Retrieval-Augmented Generation (RAG)** chatbot that answers Kerala farming questions from local text files.

- Embeddings: `sentence-transformers/all-MiniLM-L6-v2`
- Vector DB: ChromaDB (local, persistent)
- Response generation:
  - Option A: **Ollama** locally (no API key)
  - Option C fallback: rule-based grounded response from retrieved passages (works in Colab)

Malayalam support: optional translation with `Helsinki-NLP/opus-mt-ml-en` (heavy; loads on demand).

## 1) Prepare knowledge base files

The repo includes starter files under:
- `ml_models/5_chatbot/knowledge_base/`

In Colab, you can edit them or replace with your own documents.


```python
from pathlib import Path

kb_dir = Path('ml_models/5_chatbot/knowledge_base')
kb_dir.mkdir(parents=True, exist_ok=True)
print('KB files:')
for p in kb_dir.glob('*.txt'):
    print('-', p.name)
```

## 2) Build index + chat demo


```python
import importlib.util
from pathlib import Path

rag_path = Path('ml_models/5_chatbot/rag_pipeline.py')
spec = importlib.util.spec_from_file_location('rag_pipeline', rag_path)
rag = importlib.util.module_from_spec(spec)
spec.loader.exec_module(rag)

out = rag.chat(
    message='Best season to plant ginger in Wayanad and what precautions should I take?',
    language='en',
    knowledge_base_dir=kb_dir,
    chroma_dir=Path(DATA_PATH) / 'processed' / 'chroma_kb',
)
out
```
